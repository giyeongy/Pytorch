```python
import torch
```

torch.tensor(리스트)

```python
a = torch.tensor([1, 2, 3])
print(type(a))
print(a.dtype)
print(a.shape)
print(a.ndim)
print(a.numel())
```

torch.zeros(size)

```python
b = torch.zeros(3, 3, dtype=torch.int32)
print(b)
print(b.dtype)
print(b.shape)
```

torch.zeros_like(행렬)

```python
c = torch.zeros_like(a)
print(c)
print(c.dtype)
print(c.shape)
```

torch.ones(size)

```python
d = torch.ones(5, 5, dtype=torch.int64)
print(d)
print(d.dtype)
print(d.shape)
```

- torch.arange(start, end, steps): steps 간격으로 생성
- torch.linspace(start, end, steps): steps 수 만큼 생성

```python
print(torch.arange(10))
print(torch.arange(1, 10))
print(torch.arange(1, 10, 0.5))
```

```python
a = torch.linspace(1, 10, 20)
print(a)
print(a.shape)
```

- torch.randn(size): N(0, 1)
- torch.rand(size): uniform distribution
- torch.randint(low, high, size)

```python
print(torch.rand(4, 4))
print(torch.randn(4, 4))
print(torch.randint(low=1, high=10, size=(4, 4)))
```

- A*B: element-wise product(Hadamard product)
- A@B: 행렬 곱

```python
A = torch.randint(1, 10, size=(3, 3))
B = torch.randint(1, 10, size=(3, 3))
print(A+B)
print(A-B)
print(A*B)
print(A@B)
print(A/B)
print(A**2)
```

```python
X = torch.randn(3, 3)
print(X)
```

파이토치 여러 함수
- torch.abs(텐서)
- torch.exp
- torch.log
- torch.round/ceil/floor
- torch.sin/cos/tan/tanh
- torch.nan/inf/isnan/isinf
- torch.sum/mean/std
- torch.max/min
- torch.sort

```python
print(X.abs())
print(X.abs().sqrt())
print(X.exp())
print(X.log())
print(X.round())
print(X.floor())
print(X.ceil())
print(torch.sin(X))
print(torch.cos(X))
print(torch.tan(X))
print(torch.tanh(X))
print(torch.isnan(X))
print(torch.isinf(X))
```

```python
print(X)
print(X.sum())
print(X.mean())
print(X.std())
```

```python
print(X)
print(X.max())
print(X.max(dim=0))
print(X.max(dim=1))
print(X.sort(dim=0))
print(X.sort(dim=1))
```

- reshape
- transpose: 두 차원 교환(permute의 특별한 경우, 순서 상관 없음)
- permute: 모든 차원 교환(순서 중요)

```python
A = torch.rand(3, 4)
print(A.shape)
print(A.reshape(-1, 1).shape)
print(A.reshape(1, -1).shape)
```

```python
print(A.transpose(0, 1).shape)
print(A.transpose(1, 0).shape)
```

```python
print(A.permute(1, 0).shape)
print(A.permute(0, 1).shape)
```

torch.cat

```python
A = torch.rand(3, 4)
B = torch.rand(3, 4)
print(torch.cat([A, B], dim=0).shape)
print(torch.cat([A, B], dim=1).shape)
```

- torch.squeeze: dim이 1인 차원 삭제
- torch.unsqueeze: dim이 1인 차원 생성

```python
A = torch.rand(1, 1, 10)
print(A.shape)
B = torch.unsqueeze(A, dim=0)
print(B.shape)
C = torch.squeeze(B, dim=(0, 1))
print(C.shape)
```
