```python
import torch
import matplotlib.pyplot as plt
```

```python
x = torch.arange(10)
y = x**2 +1
print(x)
print(y)
```

```python
plt.scatter(x, y)
plt.show()
```

```python
# weight, bias initialization
w = torch.randn(1, requires_grad=True)
b = torch.randn(1, requires_grad=True)
print(w)
print(b)
```

```python
# forward(prediction)
y_hat = w*x + b
print(y_hat)
```

```python
# loss(MSE)
loss = ((y - y_hat)**2).mean()
```

```python
# 미분(backpropagation)
loss.backward()
```

```python
print(f'w gradient: {w.grad.item()}, b gradient: {b.grad.item()}')
```

```python
# gradient 계산 미적용
print(y_hat.requires_grad)

with torch.no_grad():
    y_hat = w*x + b

print(y_hat.requires_grad)
```
